```python
def calc_cost(row):
    if row['order_status'] == 'Cancelled':
        return 0
    elif row['order_status'] == 'Returned':
        return row['shipping_cost']
    else:
        return row['quantity'] * row['cost_price']

kp['total_cost'] = kp.apply(calc_cost, axis=1)
```

```python
kp["profit"]=kp["net_amount"]-kp["total_cost"]
```

```python
kp['profit_margin_pct'] = (kp['profit'] / kp['net_amount'] * 100).round(2)
```

```python
kp['profit_margin_pct'] = kp['profit_margin_pct'].replace([np.inf, -np.inf], np.nan)
```
